```python
dbutils.widgets.text("pipeline_name", "")
dbutils.widgets.text("table_name", "")
dbutils.widgets.text("run_id", "")
dbutils.widgets.text("rows_read", "0")
dbutils.widgets.text("rows_written", "0")
```

```python
pipeline_name = dbutils.widgets.get("pipeline_name")
table_name = dbutils.widgets.get("table_name")
run_id = dbutils.widgets.get("run_id")
rows_read = int(dbutils.widgets.get("rows_read"))
rows_written = int(dbutils.widgets.get("rows_written"))

print("Pipeline:", pipeline_name)
print("Table:", table_name)
print("Run ID:", run_id)
print("Rows read:", rows_read)
print("Rows written:", rows_written)
```

```python
spark.sql(f"""
UPDATE ecommerce.control.pipeline_audit
SET
    end_time = current_timestamp(),
    rows_read = {rows_read},
    rows_written = {rows_written},
    status = 'Succeeded',
    error_message = NULL
WHERE run_id = '{run_id}'
""")

print("Audit record updated successfully")
```

```python
%sql
SELECT
    pipeline_name,
    table_name,
    run_id,
    event_blob_path,
    start_time,
    end_time,
    rows_read,
    rows_written,
    status,
    trigger_type
FROM ecommerce.control.pipeline_audit
ORDER BY start_time DESC
LIMIT 1;
```

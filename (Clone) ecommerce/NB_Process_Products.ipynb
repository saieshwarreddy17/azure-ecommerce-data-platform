```python
x`dbutils.widgets.text("event_blob_path", "")

event_blob_path = dbutils.widgets.get("event_blob_path")

if not event_blob_path:
    raise ValueError(
        "event_blob_path is empty. This notebook must be executed from ADF."
    )

print("Event blob path:")
print(event_blob_path)
```

```python
relative_path = event_blob_path.replace("bronze/", "", 1)

bronze_path = (
    "abfss://bronze@stcommercesrc.dfs.core.windows.net/"
    + relative_path
)

print("Bronze path:")
print(bronze_path)
```

```python
print("STEP 1 - Reading Bronze product file")

df_bronze = spark.read.parquet(bronze_path)

rows_read = df_bronze.count()

print("Rows read from Bronze:", rows_read)

display(df_bronze.limit(10))
```

```python
print("Products schema:")
df_bronze.printSchema()
```

```python
display(df_bronze.limit(20))
```

```python
from pyspark.sql.functions import (
    col,
    trim,
    initcap,
    upper
)

print("STEP 2 - Cleaning product data")

df_clean = (
    df_bronze
    .withColumn("product_name", trim(col("product_name")))
    .withColumn("category_name", initcap(trim(col("category_name"))))
    .withColumn("brand", initcap(trim(col("brand"))))
    .withColumn("product_status", upper(trim(col("product_status"))))
    .withColumn("price", col("price").cast("decimal(12,2)"))
    .withColumn("cost", col("cost").cast("decimal(12,2)"))
)

print("Product cleaning completed")

display(df_clean.limit(10))
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number

print("STEP 3 - Deduplicating products")

window_product = (
    Window
    .partitionBy("product_id")
    .orderBy(col("updated_at").desc())
)

df_deduplicated = (
    df_clean
    .withColumn("rn", row_number().over(window_product))
    .filter(col("rn") == 1)
    .drop("rn")
)

rows_after_dedup = df_deduplicated.count()

print("Rows after deduplication:", rows_after_dedup)

display(df_deduplicated.limit(10))
```

```python
duplicate_count = (
    df_deduplicated
    .groupBy("product_id")
    .count()
    .filter(col("count") > 1)
    .count()
)

print("Product IDs still duplicated:", duplicate_count)

if duplicate_count != 0:
    raise ValueError(
        f"Deduplication failed. {duplicate_count} product IDs are still duplicated."
    )

print("Product duplicate check PASSED")
```

```python
from pyspark.sql.functions import (
    col,
    trim,
    when
)

print("STEP 4 - Validating product data")

df_validated = (
    df_deduplicated
    .filter(col("product_id").isNotNull())
    .filter(col("product_name").isNotNull())
    .filter(trim(col("product_name")) != "")
    .filter(col("category_id").isNotNull())
    .filter(col("price").isNotNull())
    .filter(col("price") > 0)
    .filter(col("cost").isNotNull())
    .filter(col("cost") >= 0)
    .filter(col("stock_quantity").isNotNull())
    .filter(col("stock_quantity") >= 0)
    .filter(
        col("product_status").isin(
            "ACTIVE",
            "INACTIVE",
            "DISCONTINUED"
        )
    )
)

rows_after_validation = df_validated.count()

print("Rows after deduplication:", rows_after_dedup)
print("Rows after validation:", rows_after_validation)
print(
    "Rows rejected:",
    rows_after_dedup - rows_after_validation
)

display(df_validated.limit(10))
```

```python
from delta.tables import DeltaTable

print("STEP 5 - Merging products into Silver")

silver_products = DeltaTable.forName(
    spark,
    "ecommerce.silver.products_external"
)

(
    silver_products.alias("target")
    .merge(
        df_validated.alias("source"),
        "target.product_id = source.product_id"
    )
    .whenMatchedUpdateAll()
    .whenNotMatchedInsertAll()
    .execute()
)

print("Product MERGE completed successfully")
```

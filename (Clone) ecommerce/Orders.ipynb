```python
%sql
SELECT
    order_id,
    COUNT(*) AS occurrence_count
FROM ecommerce.bronze.orders
GROUP BY order_id
HAVING COUNT(*) > 1
ORDER BY occurrence_count DESC
```

```python
%sql
SELECT
    o.customer_id,
    COUNT(*) AS order_count
FROM ecommerce.bronze.orders o
LEFT JOIN ecommerce.silver.customers c
    ON o.customer_id = c.customer_id
WHERE c.customer_id IS NULL
GROUP BY o.customer_id
ORDER BY order_count DESC;
```

```python
%sql
SELECT
    COUNT(*) AS affected_orders
FROM ecommerce.bronze.orders o
LEFT JOIN ecommerce.silver.customers c
    ON o.customer_id = c.customer_id
WHERE c.customer_id IS NULL;
```

```python
%sql
SELECT
    order_id,
    total_amount,
    discount_amount,
    tax_amount,
    final_amount,
    ROUND(
        total_amount - discount_amount + tax_amount,
        2
    ) AS expected_final_amount
FROM ecommerce.bronze.orders
WHERE ROUND(
          total_amount - discount_amount + tax_amount,
          2
      ) <> ROUND(final_amount, 2)
LIMIT 50;
```

```python
%sql
SELECT
    order_status,
    COUNT(*) AS order_count
FROM ecommerce.bronze.orders
WHERE order_status IS NULL
   OR UPPER(TRIM(order_status)) NOT IN (
       'PENDING',
       'CONFIRMED',
       'SHIPPED',
       'DELIVERED',
       'CANCELLED'
   )
GROUP BY order_status
ORDER BY order_count DESC;
```

```python
%sql
SELECT
    COUNT(*) AS invalid_amount_orders
FROM ecommerce.bronze.orders
WHERE total_amount IS NULL
   OR total_amount < 0
   OR discount_amount IS NULL
   OR discount_amount < 0
   OR tax_amount IS NULL
   OR tax_amount < 0
   OR final_amount IS NULL
   OR final_amount < 0;
```

```python
%sql
SELECT
    COUNT(*) AS invalid_date_orders
FROM ecommerce.bronze.orders
WHERE order_date IS NULL
   OR created_at IS NULL
   OR updated_at IS NULL
   OR updated_at < created_at
   OR order_date > CURRENT_TIMESTAMP();
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.orders
USING DELTA
AS
SELECT
    order_id,
    customer_id,
    order_date,
    INITCAP(TRIM(order_status)) AS order_status,
    INITCAP(TRIM(payment_status)) AS payment_status,
    INITCAP(TRIM(shipping_status)) AS shipping_status,
    shipping_address_id,
    CAST(total_amount AS DECIMAL(12,2)) AS total_amount,
    CAST(discount_amount AS DECIMAL(12,2)) AS discount_amount,
    CAST(tax_amount AS DECIMAL(12,2)) AS tax_amount,
    CAST(final_amount AS DECIMAL(12,2)) AS final_amount,
    created_at,
    updated_at
FROM ecommerce.bronze.orders
WHERE order_id IS NOT NULL
  AND order_date IS NOT NULL
  AND created_at IS NOT NULL
  AND updated_at IS NOT NULL
  AND updated_at >= created_at
  AND order_date <= CURRENT_TIMESTAMP()
  AND total_amount IS NOT NULL
  AND total_amount >= 0
  AND discount_amount IS NOT NULL
  AND discount_amount >= 0
  AND tax_amount IS NOT NULL
  AND tax_amount >= 0
  AND final_amount IS NOT NULL
  AND final_amount >= 0
  AND UPPER(TRIM(order_status)) IN (
      'PLACED',
      'PENDING',
      'CONFIRMED',
      'SHIPPED',
      'DELIVERED',
      'CANCELLED'
  );
```

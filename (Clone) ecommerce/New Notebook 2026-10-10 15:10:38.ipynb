```python
from pyspark.sql import functions as F

bronze_path = "abfss://bronze@stcommercesrc.dfs.core.windows.net/customer_addresses/"

df_addresses = spark.read.parquet(bronze_path)

df_silver_addresses = (
    df_addresses
    .withColumn("is_default", F.col("is_default").cast("boolean"))
    .withColumn(
        "is_effective_default",
        F.col("is_default")
    )
    .withColumn("created_at", F.col("created_at").cast("timestamp"))
    .withColumn("updated_at", F.col("updated_at").cast("timestamp"))
)

(
    df_silver_addresses.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("ecommerce.silver.customer_addresses")
)
```

```python
%sql
INSERT INTO ecommerce.silver.customers
    (customer_id, first_name, last_name, email, phone, gender,
     date_of_birth, registration_date, customer_status, created_at, updated_at)
WITH ranked_customers AS (
    SELECT
        customer_id,
        first_name,
        last_name,
        email,
        phone,
        gender,
        date_of_birth,
        registration_date,
        customer_status,
        created_at,
        updated_at,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id
            ORDER BY updated_at DESC, created_at DESC
        ) AS rn
    FROM ecommerce.bronze.customers
    WHERE customer_id IN (107, 108, 109, 110)
)
SELECT
    b.customer_id,
    b.first_name,
    b.last_name,
    b.email,
    b.phone,
    b.gender,
    b.date_of_birth,
    b.registration_date,
    b.customer_status,
    b.created_at,
    b.updated_at
FROM ranked_customers b
WHERE b.rn = 1
  AND NOT EXISTS (
      SELECT 1
      FROM ecommerce.silver.customers s
      WHERE s.customer_id = b.customer_id
  );
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.gold.regional_sales
USING DELTA
LOCATION 'abfss://gold@stcommercesrc.dfs.core.windows.net/regional_sales/'
AS
SELECT
    a.state AS state,
    a.city AS city,
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(SUM(o.final_amount), 2) AS total_sales,
    ROUND(AVG(o.final_amount), 2) AS average_order_value
FROM ecommerce.silver.orders o
JOIN ecommerce.silver.customers c
    ON o.customer_id = c.customer_id
JOIN ecommerce.silver.customer_addresses a
    ON c.customer_id = a.customer_id
    AND a.is_effective_default = true
GROUP BY
    a.state,
    a.city;
```

```python
%sql
SELECT
    state,
    SUM(total_orders) AS total_orders,
    ROUND(SUM(total_sales), 2) AS total_sales
FROM ecommerce.gold.regional_sales
GROUP BY state
ORDER BY total_sales DESC;
```

```python
%sql
SELECT
    SUM(total_orders) AS total_orders,
    ROUND(SUM(total_sales), 2) AS total_sales
FROM ecommerce.gold.regional_sales;
```

```python
%sql
SELECT
    MAX(segmentation_timestamp) AS segmentation_date,
    MIN(recency_days) AS min_recency_days,
    MAX(recency_days) AS max_recency_days,
    COUNT(*) AS total_customers
FROM ecommerce.gold.customer_segmentation;
```

```python
%sql
SELECT
    customer_segment,
    COUNT(*) AS customer_count,
    SUM(CASE WHEN total_orders = 0 THEN 1 ELSE 0 END) AS zero_order_customers,
    SUM(CASE WHEN last_order_date IS NULL THEN 1 ELSE 0 END) AS missing_last_order_date,
    MIN(recency_days) AS min_recency_days,
    MAX(recency_days) AS max_recency_days
FROM ecommerce.gold.customer_segmentation
GROUP BY customer_segment
ORDER BY customer_count DESC;
```

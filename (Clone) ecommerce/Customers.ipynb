**CREATING SILVER TABLE**

```python
%sql
CREATE SCHEMA IF NOT EXISTS ecommerce.silver
COMMENT 'Silver schema containing cleaned and deduplicated customer data';
```

```python
%sql
SHOW SCHEMAS IN ecommerce;
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.customers
USING DELTA
AS
WITH ranked_customers AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id
            ORDER BY updated_at DESC
        ) AS rn
    FROM ecommerce.bronze.customers
)
SELECT
    customer_id,
    first_name,
    last_name,
    email,
    phone,
    gender,
    date_of_birth,
    registration_date,
    customer_status,
    created_at,
    updated_at
FROM ranked_customers
WHERE rn = 1;
```

**CLEANING FOR THE SILVER**

```python
%sql
SELECT
    customer_id,
    first_name,
    last_name,
    email,
    gender,
    customer_status
FROM ecommerce.silver.customers
WHERE
    first_name <> TRIM(first_name)
    OR last_name <> TRIM(last_name)
    OR email <> LOWER(TRIM(email))
    OR customer_status <> INITCAP(TRIM(customer_status))
ORDER BY customer_id;
```

**Re-Build for silver table**

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.customers
USING DELTA
AS
WITH ranked_customers AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id
            ORDER BY updated_at DESC
        ) AS rn
    FROM ecommerce.bronze.customers
)
SELECT
    customer_id,

    TRIM(first_name) AS first_name,

    TRIM(last_name) AS last_name,

    LOWER(TRIM(email)) AS email,

    TRIM(phone) AS phone,

    INITCAP(TRIM(gender)) AS gender,

    date_of_birth,

    registration_date,

    INITCAP(TRIM(customer_status)) AS customer_status,

    created_at,
    updated_at

FROM ranked_customers
WHERE rn = 1;
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.customer_quality_summary
USING DELTA
AS

SELECT
    'NULL_EMAIL' AS quality_rule,
    COUNT(*) AS failed_records,
    'Customer email is missing' AS description
FROM ecommerce.silver.customers
WHERE email IS NULL

UNION ALL

SELECT
    'UNKNOWN_STATUS' AS quality_rule,
    COUNT(*) AS failed_records,
    'Customer status is UNKNOWN' AS description
FROM ecommerce.silver.customers
WHERE customer_status = 'UNKNOWN';
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.customer_quality_summary
USING DELTA
AS

SELECT
    'NULL_EMAIL' AS quality_rule,
    COUNT(*) AS failed_records,
    'Customer email is missing' AS description
FROM ecommerce.silver.customers
WHERE email IS NULL

UNION ALL

SELECT
    'UNKNOWN_STATUS' AS quality_rule,
    COUNT(*) AS failed_records,
    'Customer status is UNKNOWN/Unknown' AS description
FROM ecommerce.silver.customers
WHERE LOWER(TRIM(customer_status)) = 'unknown';
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.customer_quarantine
USING DELTA
AS
SELECT
    *,
    CASE
        WHEN email IS NULL
             AND LOWER(TRIM(customer_status)) = 'unknown'
            THEN 'NULL_EMAIL_AND_UNKNOWN_STATUS'

        WHEN email IS NULL
            THEN 'NULL_EMAIL'

        WHEN LOWER(TRIM(customer_status)) = 'unknown'
            THEN 'UNKNOWN_STATUS'

        ELSE 'UNKNOWN'
    END AS rejection_reason
FROM ecommerce.silver.customers
WHERE
    email IS NULL
    OR LOWER(TRIM(customer_status)) = 'unknown';
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.customers
USING DELTA
AS
WITH ranked_customers AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id
            ORDER BY updated_at DESC
        ) AS rn
    FROM ecommerce.bronze.customers
),
cleaned_customers AS (
    SELECT
        customer_id,
        TRIM(first_name) AS first_name,
        TRIM(last_name) AS last_name,
        LOWER(TRIM(email)) AS email,
        TRIM(phone) AS phone,
        INITCAP(TRIM(gender)) AS gender,
        date_of_birth,
        registration_date,
        INITCAP(TRIM(customer_status)) AS customer_status,
        created_at,
        updated_at
    FROM ranked_customers
    WHERE rn = 1
)
SELECT *
FROM cleaned_customers
WHERE email IS NOT NULL
  AND LOWER(TRIM(customer_status)) <> 'unknown';
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.customer_quality_metrics
USING DELTA
AS
WITH ranked AS (
    SELECT
        *,
        ROW_NUMBER() OVER (
            PARTITION BY customer_id
            ORDER BY updated_at DESC
        ) AS rn
    FROM ecommerce.bronze.customers
),

latest_customers AS (
    SELECT *
    FROM ranked
    WHERE rn = 1
),

quality_classification AS (
    SELECT
        customer_id,
        CASE
            WHEN email IS NULL
                 OR LOWER(TRIM(customer_status)) = 'unknown'
            THEN 'REJECTED'
            ELSE 'PASSED'
        END AS quality_status
    FROM latest_customers
)

SELECT
    CURRENT_TIMESTAMP() AS quality_check_timestamp,

    COUNT(*) AS total_records,

    SUM(
        CASE
            WHEN quality_status = 'PASSED' THEN 1
            ELSE 0
        END
    ) AS passed_records,

    SUM(
        CASE
            WHEN quality_status = 'REJECTED' THEN 1
            ELSE 0
        END
    ) AS rejected_records,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN quality_status = 'PASSED' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS quality_percentage

FROM quality_classification;
```

```python
%sql
SELECT *
FROM ecommerce.silver.customer_quality_metrics;
```

```python
%sql
SELECT
    total_records,
    passed_records,
    rejected_records,
    total_records - (passed_records + rejected_records) AS unexplained_records,
    quality_percentage
FROM ecommerce.silver.customer_quality_metrics;
```

```python
%sql
SELECT
    customer_id,
    date_of_birth,
    registration_date,
    customer_status
FROM ecommerce.silver.customers
WHERE
    date_of_birth > CURRENT_DATE()
    OR registration_date > CURRENT_TIMESTAMP()
    OR LOWER(TRIM(customer_status)) NOT IN (
        'active',
        'inactive',
        'blocked',
        'suspended'
    )
ORDER BY customer_id;
```

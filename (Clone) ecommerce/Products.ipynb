```python
%sql
CREATE TABLE ecommerce.bronze.products
USING PARQUET
LOCATION 'abfss://bronze@stcommercesrc.dfs.core.windows.net/products/';
```

```python
%sql
SELECT COUNT(*) AS product_count
FROM ecommerce.bronze.products;
```

```python
%sql
DESCRIBE TABLE ecommerce.bronze.products;
```

```python
%sql
SELECT *
FROM ecommerce.bronze.products
LIMIT 10;
```

```python
%sql
SELECT
    product_id,
    COUNT(*) AS occurrence_count
FROM ecommerce.bronze.products
GROUP BY product_id
HAVING COUNT(*) > 1
ORDER BY occurrence_count DESC;
```

```python
%sql
SELECT
    COUNT(*) AS total_records,

    SUM(
        CASE
            WHEN product_name IS NULL
                 OR TRIM(product_name) = ''
            THEN 1 ELSE 0
        END
    ) AS invalid_product_name,

    SUM(
        CASE
            WHEN price IS NULL
                 OR price <= 0
            THEN 1 ELSE 0
        END
    ) AS invalid_price,

    SUM(
        CASE
            WHEN cost IS NULL
                 OR cost < 0
            THEN 1 ELSE 0
        END
    ) AS invalid_cost,

    SUM(
        CASE
            WHEN stock_quantity IS NULL
                 OR stock_quantity < 0
            THEN 1 ELSE 0
        END
    ) AS invalid_stock,

    SUM(
        CASE
            WHEN product_status IS NULL
                 OR UPPER(TRIM(product_status))
                    NOT IN ('ACTIVE', 'INACTIVE')
            THEN 1 ELSE 0
        END
    ) AS invalid_status

FROM ecommerce.bronze.products;
```

```python
%sql
SELECT
    product_status,
    COUNT(*) AS record_count
FROM ecommerce.bronze.products
WHERE product_status IS NULL
   OR UPPER(TRIM(product_status))
      NOT IN ('ACTIVE', 'INACTIVE')
GROUP BY product_status
ORDER BY record_count DESC;
```

```python
%sql
SELECT
    COUNT(*) AS invalid_status_count
FROM ecommerce.bronze.products
WHERE product_status IS NULL
   OR UPPER(TRIM(product_status))
      NOT IN ('ACTIVE', 'INACTIVE', 'DISCONTINUED');
```

```python
%sql
CREATE OR REPLACE TABLE ecommerce.silver.products
USING DELTA
AS
SELECT
    product_id,

    TRIM(product_name) AS product_name,

    category_id,

    INITCAP(TRIM(category_name)) AS category_name,

    INITCAP(TRIM(brand)) AS brand,

    CAST(price AS DECIMAL(12,2)) AS price,

    CAST(cost AS DECIMAL(12,2)) AS cost,

    stock_quantity,

    INITCAP(TRIM(product_status)) AS product_status,

    created_at,

    updated_at

FROM ecommerce.bronze.products
WHERE product_name IS NOT NULL
  AND TRIM(product_name) <> ''
  AND price IS NOT NULL
  AND price > 0
  AND cost IS NOT NULL
  AND cost >= 0
  AND stock_quantity IS NOT NULL
  AND stock_quantity >= 0
  AND UPPER(TRIM(product_status))
      IN ('ACTIVE', 'INACTIVE', 'DISCONTINUED');
```

```python
%sql
SELECT COUNT(*) AS silver_product_count
FROM ecommerce.silver.products;
```

```python
%sql
DESCRIBE TABLE ecommerce.silver.products;
```

```python
%sql
SELECT COUNT(*) AS product_count
FROM ecommerce.silver.products_external;
```
